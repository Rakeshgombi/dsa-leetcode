# Reverse Linked List

Given the `head` of a singly linked list, reverse the list, and return _the reversed list_.

### Example 1:

![](https://assets.leetcode.com/uploads/2021/02/19/rev1ex1.jpg)

**Input:** head = [1,2,3,4,5]

**Output:** [5,4,3,2,1]

### Example 2:

![](https://assets.leetcode.com/uploads/2021/02/19/rev1ex2.jpg)

**Input:** head = [1,2]

**Output:** [2,1]

### Example 3:

**Input:** head = []

**Output:** []

### Constraints:

*   The number of nodes in the list is the range `[0, 5000]`.
*   `-5000 <= Node.val <= 5000`

**Follow up:** A linked list can be reversed either iteratively or recursively. Could you implement both?

````python
from typing import Optional


# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        prev: ListNode = None
        curr: ListNode = head

        while curr:
            curr.next, prev, curr = prev, curr, curr.next
        return prev


soln = Solution()
resp = soln.reverseList(ListNode(1, ListNode(2, ListNode(3, ListNode(4, ListNode(5))))))
while resp is not None:
    print(resp.val)
    resp = resp.next
````

## Visualizing the Iterative Reversal

We walk three pointers through the list: `prev`, `curr`, and `temp` (saved next node). At each step we flip `curr.next` to point backward, then slide all pointers forward.

### Original list

```
head
 │
 ▼
[1] ──▶ [2] ──▶ [3] ──▶ [4] ──▶ [5] ──▶ None
```

### Step-by-step trace

```
prev = None,  curr = [1]

Step 1:  temp = [2]
         None ◀── [1]     [2] ──▶ [3] ──▶ [4] ──▶ [5] ──▶ None
                  ▲        ▲
                 prev     curr

Step 2:  temp = [3]
         None ◀── [1] ◀── [2]     [3] ──▶ [4] ──▶ [5] ──▶ None
                          ▲        ▲
                         prev     curr

Step 3:  temp = [4]
         None ◀── [1] ◀── [2] ◀── [3]     [4] ──▶ [5] ──▶ None
                                  ▲        ▲
                                 prev     curr

Step 4:  temp = [5]
         None ◀── [1] ◀── [2] ◀── [3] ◀── [4]     [5] ──▶ None
                                          ▲        ▲
                                         prev     curr

Step 5:  temp = None
         None ◀── [1] ◀── [2] ◀── [3] ◀── [4] ◀── [5]     None
                                                  ▲        ▲
                                                 prev     curr

curr = None  →  loop ends, return prev
```

### Reversed list

```
head
 │
 ▼
[5] ──▶ [4] ──▶ [3] ──▶ [2] ──▶ [1] ──▶ None
```

### Pointer movement each iteration

| Iteration | `curr` | `temp = curr.next` | `curr.next = prev` | `prev` becomes | `curr` becomes |
|:---------:|:------:|:------------------:|:------------------:|:--------------:|:--------------:|
| 1 | 1 | 2 | 1 → None | 1 | 2 |
| 2 | 2 | 3 | 2 → 1 | 2 | 3 |
| 3 | 3 | 4 | 3 → 2 | 3 | 4 |
| 4 | 4 | 5 | 4 → 3 | 4 | 5 |
| 5 | 5 | None | 5 → 4 | 5 | None |

**Time:** $O(n)$ &nbsp;&nbsp; **Space:** $O(1)$
